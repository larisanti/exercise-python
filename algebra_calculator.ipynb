### Proportion (Regra de Três)

```python
def resolver_proporcao(n1, d1, n2, d2):
    """
    Resolve uma regra de três simples. 
    Valor 0 é o que desejo descobrir (incógnita).
    """
    answer_variable = ""
    
    # Se n2 for a incógnita
    if n2 == 0:
        answer = d2 * n1 / d1
        n2 = answer 
        answer_variable = "n2"

    # Se d2 for a incógnita
    elif d2 == 0:
        answer = n2 * d1 / n1
        d2 = answer
        answer_variable = "d2"
        
    # Se n1 for a incógnita
    elif n1 == 0:
        answer = d1 * n2 / d2
        n1 = answer
        answer_variable = "n1"
        
    # Se d1 for a incógnita
    elif d1 == 0:
        answer = n1 * d2 / n2
        d1 = answer
        answer_variable = "d1"

    if answer_variable:
        print(f"Resultado: {answer_variable} = {answer}")
        print(f"Proporção completa: {n1}/{d1} = {n2}/{d2}")
    else:
        print("Nenhuma incógnita (0) foi informada.")
```

```python
# Resolver (icógnita = 0)
resolver_proporcao(n1=1, d1=2, n2=4, d2=0)
```

### Média de Homework Score

Automatizei o cálculo da média do aproveitamento de homework dos meus alunos.

```python
score_total = []
qtd_atividade = int(input("Quantidade de atividades:"))

# Loop para calcular o score de cada atividade com base nos acertos e total
for i in range(1, qtd_atividade + 1):
    entrada = input(f"Atividade {i} (acerto/total): ")
    
    a, t = map(int, entrada.split('/'))
    score_parcial = (a * 100) / t
    
    score_total.append(score_parcial)
    print(f"Atividade {i}: {score_parcial:.2f}%")

# Média simples para score total
if score_total:
    media_final = sum(score_total) / len(score_total)
    print(f"Score: {media_final:.2f}%")
```
